```python
import pandas as pd
import json

df = pd.read_json('model-training/data/reviews_carsurvej.json')

df_parsed = df[0].apply(json.loads).apply(pd.Series)
df_parsed
```

```python
df_parsed.isna().any()
```

```python
df_parsed[df_parsed['Summary:'].isna()]
```

```python
df_parsed_filled = df_parsed[~df_parsed['General Comments:'].isna()]
```

```python
df_parsed_filled['Faults:'].fillna('Without faults', inplace=True)
df_parsed_filled['Summary:'].fillna('Stock car', inplace=True)
```

```python
df_parsed_filled.isna().any()
```

```python
df_restored = df_parsed.apply(lambda x: json.dumps(x.to_dict()), axis=1).to_frame()
df_restored.to_json('restored_reviews.json', orient='records', lines=True)  # or without 'lines=True' for array format
```

```python
with open('restored_reviews_carsurvej.json', 'w', encoding='utf-8') as f:
    json.dump(df_parsed.to_dict(orient='records'), f, indent=2, ensure_ascii=False)
```
